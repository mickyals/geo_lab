```python
import xarray as xr
import matplotlib.pyplot as plt
import imageio
import os

data_dir = r"C:\Users\micke\Downloads\dfda178b22d3772b9b9b0118dba68a02.nc"
data = xr.open_dataset(data_dir)
data["longitude"] = (data["longitude"]+180) % 360 -180

data = data.sortby("longitude")
data
```

```python
# -------------------------------
# User settings
# -------------------------------

                      # your xarray dataset
var = 't'        # <-- specify variable to plot
time_dim = "valid_time"                 # or "valid_time", depending on dataset
pressure_dim = "pressure_level"
lat_dim = "latitude"
lon_dim = "longitude"

# -------------------------------
# Output directory setup
# -------------------------------

top_dir = f"./{var}"
os.makedirs(top_dir, exist_ok=True)

# -------------------------------
# Loop over pressure levels
# -------------------------------

pressure_levels = data[pressure_dim].values

for p in pressure_levels:

    sub_dir = os.path.join(top_dir, f"{pressure_dim}_{int(p)}")
    os.makedirs(sub_dir, exist_ok=True)

    frames = []   # store file paths for GIF assembly

    data_p = data[var].sel({pressure_dim: p})

    # Iterate through each time step
    for idx, t in enumerate(data_p[time_dim].values):

        da_plot = data_p.sel({time_dim: t})

        fig, ax = plt.subplots(figsize=(10, 5))
        im = da_plot.plot(ax=ax, cmap="viridis", add_colorbar=True)

        ax.set_title(f"{var} — {pressure_dim}={p} — time={str(t)}")

        # Save image
        frame_path = os.path.join(sub_dir, f"frame_{idx:03d}.png")
        plt.savefig(frame_path, dpi=150, bbox_inches="tight")
        plt.close(fig)

        frames.append(frame_path)

    # -------------------------------
    # Build GIF for this pressure level
    # -------------------------------
    gif_path = os.path.join(sub_dir, f"{var}_{int(p)}.gif")

    with imageio.get_writer(gif_path, mode="I", duration=0.3) as writer:
        for frame in frames:
            writer.append_data(imageio.imread(frame))

    print(f"Saved GIF for pressure level {p}: {gif_path}")
```
