```python
import xarray as xr
import numpy as np
import json
from pathlib import Path

def convert_netcdf_to_npy(
    source_dir: str,
    dest_dir: str,
    variables: list[str]
):
    source_path = Path(source_dir)
    dest_path = Path(dest_dir)
    dest_path.mkdir(parents=True, exist_ok=True)

    print(f"Opening NetCDFs from {source_path}...")
    if source_path.is_file():
        ds = xr.open_dataset(source_path)
    else:
        ds = xr.open_mfdataset(str(source_path / "*.nc"), combine='by_coords')

    # Helper function to convert numpy scalars to python natives for JSON
    def get_stats(arr):
        return {
            "minimum": float(arr.min()),
            "maximum": float(arr.max()),
            "mean": float(arr.mean()),
            "std": float(arr.std())
        }

    # 1. Process Coordinates
    print("Processing coordinates...")
    coords = {}
    coord_stats = {}

    # Time
    coords['valid_time'] = ds['valid_time'].values.astype('datetime64[s]').astype(np.float32)
    coord_stats['valid_time'] = get_stats(coords['valid_time'])

    # Pressure
    coords['pressure_level'] = ds['pressure_level'].values.astype(np.float32)
    coord_stats['pressure_level'] = get_stats(coords['pressure_level'])

    # Latitude
    coords['latitude'] = ds['latitude'].values.astype(np.float32)
    coord_stats['latitude'] = get_stats(coords['latitude'])

    # Longitude
    lon_data = ds['longitude'].values
    lon_data = ((lon_data + 180) % 360) - 180
    coords['longitude'] = lon_data.astype(np.float32)
    coord_stats['longitude'] = get_stats(coords['longitude'])

    # Save Coordinates
    np.savez(dest_path / "coords.npz", **coords)

    # 2. Process Variables
    shape_info = {}
    var_stats = {}
    for var in variables:
        print(f"Processing variable: {var}...")
        if var not in ds:
            continue

        data = ds[var].values.astype(np.float32)
        var_stats[var] = get_stats(data)

        np.save(dest_path / f"{var}.npy", data)
        # Convert shape tuple to list of ints for JSON safety
        shape_info[var] = [int(s) for s in data.shape]

    # 3. Save Metadata
    metadata = {
        "coord_labels": {k: i for i, k in enumerate(["valid_time", "pressure_level", "latitude", "longitude"])},
        "variable_labels": {k: i for i, k in enumerate(variables)},
        "coord_stats": coord_stats,
        "var_stats": var_stats,
        "shapes": shape_info
    }

    with open(dest_path / "metadata.json", "w") as f:
        json.dump(metadata, f, indent=2)

    print(f"Conversion complete. Data saved to {dest_path}")

convert_netcdf_to_npy(data_dir, "./era5_npy", ["u", "v", "t", "z", "w"])
```

```python
# Cell 1: Imports
import numpy as np
import torch
from torch.utils.data import DataLoader, random_split
from geolab.data.dataset import ERA5MultiData, ERA5MultiDataset

# Cell 2: Initialize ERA5MultiData (you've already done this, but keeping it for completeness)
data_dir = r"C:\Users\micke\Downloads\dfda178b22d3772b9b9b0118dba68a02.nc"
variables = ['t', 'w', 'u', 'z', 'v']

# Load a small subset for testing
era5_data = ERA5MultiData(
    data_dir=data_dir,
    variables=variables,
    time_idx_range=[0, 10],      # First 10 time steps
    pressure_idx_range=[0, 5],   # First 5 pressure levels
    latitude_idx_range=[0, 100], # First 100 latitude points
    longitude_idx_range=[0, 100], # First 100 longitude points
    preload=True
)

# Cell 3: Create indices for the dataset
# This creates a grid of all possible combinations of indices
time_indices = np.arange(era5_data.num_times)
pressure_indices = np.arange(era5_data.num_pressure_levels)
lat_indices = np.arange(era5_data.num_latitudes)
lon_indices = np.arange(era5_data.num_longitudes)

# Create a meshgrid of all combinations
grid = np.meshgrid(time_indices, pressure_indices, lat_indices, lon_indices, indexing='ij')
indices = np.stack([g.ravel() for g in grid], axis=1)  # Shape: (N, 4)

print(f"Created {len(indices)} data points")

# Cell 4: Split into train/val/test
# For demonstration, let's use 70% train, 15% val, 15% test
n_total = len(indices)
train_size = int(0.7 * n_total)
val_size = int(0.15 * n_total)
test_size = n_total - train_size - val_size

# Split indices
train_indices, val_indices, test_indices = random_split(
    range(n_total), [train_size, val_size, test_size],
    generator=torch.Generator().manual_seed(42)
)

# Create datasets
train_dataset = ERA5MultiDataset(
    data=era5_data,
    indices=indices[train_indices.indices],
    statistics={},  # Not used in current implementation
    variables=variables
)

val_dataset = ERA5MultiDataset(
    data=era5_data,
    indices=indices[val_indices.indices],
    statistics={},
    variables=variables
)

test_dataset = ERA5MultiDataset(
    data=era5_data,
    indices=indices[test_indices.indices],
    statistics={},
    variables=variables
)

print(f"Train samples: {len(train_dataset)}")
print(f"Val samples: {len(val_dataset)}")
print(f"Test samples: {len(test_dataset)}")

# Cell 5: Create DataLoaders
batch_size = 32
num_workers = 4  # Adjust based on your CPU cores

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=num_workers,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=num_workers,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=num_workers,
    pin_memory=True
)

# Cell 6: Test the DataLoader
# Get a batch of data
sample_batch = next(iter(train_loader))
coords = sample_batch['coords']  # Shape: [batch_size, 4]
values = sample_batch['values']  # Shape: [batch_size, num_vars]

print("\nSample batch:")
print(f"Coordinates shape: {coords.shape}")
print(f"Values shape: {values.shape}")
print("\nFirst coordinate batch:")
print(coords[0])
print("\nCorresponding values:")
print(values[0])
```

```python
# Cell 1: Imports and setup
import torch
from geolab.data.datamodule import TroposphereDataModule

# Cell 2: Initialize the datamodule with your data
data_dir = r"C:\Users\micke\Downloads\dfda178b22d3772b9b9b0118dba68a02.nc"
variables = ['t', 'w', 'u', 'z', 'v']

datamodule = TroposphereDataModule(
    data_dir=data_dir,
    solution_vars=variables,
    time_idx_range=[0, 10],       # First 10 time steps
    pressure_idx_range=[0, 5],    # First 5 pressure levels
    latitude_idx_range=[0, 100],  # First 100 latitude points
    longitude_idx_range=[0, 100], # First 100 longitude points
    batch_size=32,
    num_workers=4,
    val_split=0.15,
    test_split=0.15,
    norm_type="default",  # Options: "default" (range [-1,1]), "standard" (z-score), or None
    seed=42
)

# Cell 3: Setup the datamodule (this will load data and compute statistics)
datamodule.prepare_data()  # Validates data exists
datamodule.setup(stage="fit")  # Loads data and computes statistics

# Cell 4: Check data statistics
print("\nComputed statistics:")
for var in variables:
    stats = datamodule.statistics[var]
    print(f"{var}:")
    print(f"  Min: {stats['min'].item():.4f}")
    print(f"  Max: {stats['max'].item():.4f}")
    print(f"  Mean: {stats['mean'].item():.4f}")
    print(f"  Std: {stats['std'].item():.4f}")

# Cell 5: Get a batch of data and check normalization
train_loader = datamodule.train_dataloader()
batch = next(iter(train_loader))

# Original values
coords = batch['coords']  # Shape: [batch_size, 4]
values = batch['values']  # Shape: [batch_size, num_vars]

print("\nSample batch before normalization:")
print("Coordinates shape:", coords.shape)
print("Values shape:", values.shape)
print("\nFirst coordinate (raw):", coords[0])
print("First values (raw):", values[0])

# Normalize the batch
normalized_batch = datamodule.normalize_batch(batch)
norm_coords = normalized_batch['coords']
norm_values = normalized_batch['values']

print("\nAfter normalization:")
print("Normalized coordinates (should be in ~[-1,1]):", norm_coords[0])
print("Normalized values (should be in ~[-1,1]):", norm_values[0])

# Check normalization ranges
print("\nNormalization check:")
print("  Coords min/max:", norm_coords.min().item(), norm_coords.max().item())
for i, var in enumerate(variables):
    var_values = norm_values[:, i]
    print(f"  {var} min/max: {var_values.min().item():.4f} / {var_values.max().item():.4f}")

# Cell 6: Test denormalization
denorm_batch = {
    'coords': norm_coords.clone(),
    'values': norm_values.clone()
}
denormalized = datamodule.denormalize_batch(denorm_batch)

# Check if denormalization recovers original values
coord_diff = torch.abs(coords - denormalized['coords']).max().item()
values_diff = torch.abs(values - denormalized['values']).max().item()

print("\nDenormalization check:")
print(f"  Max coordinate difference: {coord_diff:.2e} (should be close to 0)")
print(f"  Max values difference: {values_diff:.2e} (should be close to 0)")

# Cell 7: Test with a DataLoader
print("\nTesting with DataLoader...")
for i, batch in enumerate(train_loader):
    if i >= 2:  # Just check first 2 batches
        break

    # Normalize the batch
    norm_batch = datamodule.normalize_batch(batch)
    coords, values = norm_batch['coords'], norm_batch['values']

    print(f"\nBatch {i+1}:")
    print("  Coords range: [{:.4f}, {:.4f}]".format(
        coords.min().item(), coords.max().item()
    ))

    for j, var in enumerate(variables):
        var_values = values[:, j]
        print(f"  {var} range: [{var_values.min().item():.4f}, {var_values.max().item():.4f}]")
```
